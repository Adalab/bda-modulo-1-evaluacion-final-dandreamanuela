```python
# 2.6 calcular_valor_inventario(): Calcula y muestra el valor total del inventario.

# Utiliza un bucle for para calcular el valor total del inventario.
# Itera a través del inventario y suma el valor de cada producto (precio x cantidad) -> calcula el valor total del inventario 

inventario = [
{'nombre': 'Camisa', 'precio': 20, 'cantidad': 40},
{'nombre': 'Pantalón', 'precio': 30, 'cantidad': 30},
{'nombre': 'Zapatos', 'precio': 50, 'cantidad': 25},
{'nombre': 'Chaqueta', 'precio': 70, 'cantidad': 15},
{'nombre': 'Falda', 'precio': 25, 'cantidad': 20},
{'nombre': 'Gorra', 'precio': 10, 'cantidad': 50},
{'nombre': 'Bufanda', 'precio': 15, 'cantidad': 35},
{'nombre': 'Calcetines', 'precio': 5, 'cantidad': 100},
{'nombre': 'Cinturón', 'precio': 12, 'cantidad': 40},
{'nombre': 'Bolso', 'precio': 45, 'cantidad': 18}
]

def calcular_valor_inventario():
    valor_total = 0
    for producto in inventario:
        valor_producto = producto["precio"] * producto["cantidad"]
        valor_total = valor_total + valor_producto
    return valor_total

print(calcular_valor_inventario())
```

```python
#2.7 realizar_compra(): Permite a un cliente realizar una compra seleccionando productos del inventario

# Debe interactuar con el cliente para seleccionar productos y calcular el costo total de la compra
# Utiliza un bucle while para permitir al cliente realizar múltiples compras.
# Muestra el inventario y solicita al cliente ingresar el nombre del producto que desea comprar.
# Registra los productos seleccionados en un carrito y actualiza el inventario.
# Calcula el costo total de la compra.

def realizar_compra():
    carrito = []
    compra_total = 0

    while True:
        producto_seleccionado = input("Indica el producto que deseas comprar o escribe 'fin' para terminar: ")

        if producto_seleccionado.lower() == "fin":
            break

        encontrado = False
        
        for producto in inventario:
            if producto_seleccionado.lower() == producto["nombre"].lower():
                encontrado = True
                
                print(producto["nombre"],"- stock disponible: ", producto["cantidad"])

                cantidad_seleccionada = int(input("Indica la cantidad de producto que deseas comprar: "))

                if producto["cantidad"] == 0:
                    print("El producto está agotado")

                elif cantidad_seleccionada <= producto["cantidad"]:
                    producto["cantidad"] = producto["cantidad"] - cantidad_seleccionada

                    compra_total = compra_total + (producto["precio"] * cantidad_seleccionada)

                    carrito.append({
                        "nombre": producto["nombre"],
                        "cantidad": cantidad_seleccionada,
                        "precio": (producto["precio"] * cantidad_seleccionada)
                    })

                    print("Producto", producto["nombre"], "añadido al carrito")

                else:
                    print("No hay suficiente stock")
                break

        if encontrado == False:
            print("Producto no encontrado")
   

    print("\n--- RESUMEN DE COMPRA ---")
    for item in carrito:
        print(item["cantidad"], "x", item["nombre"], "=", "Precio total: $", item["precio"])

    print("Total a pagar: $", compra_total)

realizar_compra()
```
